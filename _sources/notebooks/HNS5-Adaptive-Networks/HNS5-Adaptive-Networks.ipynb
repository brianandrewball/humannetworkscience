# Adaptive networks

## What are adaptive networks?

Adaptive networks are a type of network whose structure evolves in response to the states and dynamics of their nodes. Unlike static networks, where connections between nodes are fixed, or purely temporal networks, which change over time without necessarily being influenced by node properties, adaptive networks exhibit a dynamic interplay between node states and network topology. This mutual feedback mechanism makes adaptive networks particularly suitable for modeling systems where both the interactions and the individual components evolve simultaneously.


In adaptive networks, connections (or *edges*) can change in response to node properties such as states, behaviors, or other intrinsic factors. For instance, in social networks, friendships may form or dissolve depending on shared beliefs or changing opinions, leading to a network structure that reflects the evolving preferences and attitudes of individuals. Similarly, in ecological systems, the interactions between species—such as predator-prey relationships—can adapt in response to population dynamics, environmental changes, and resource availability. These networks thus enable a more nuanced understanding of complex systems, where both individual behavior and network structure influence one another continuously.

The concept of adaptive networks has been explored extensively in various fields. In epidemiology, for example, adaptive networks have been used to model how the spread of infectious diseases is influenced by adaptive social distancing behaviors, where individuals sever connections to infected nodes to protect themselves (Gross et al., 2006). In sociology, adaptive networks are used to understand phenomena such as opinion formation and polarization, where individuals adjust their social ties based on agreement or disagreement over certain issues (Holme & Newman, 2012). Ecological studies also utilize adaptive networks to model the resilience of ecosystems, where species interactions adapt in response to external pressures like climate change (Lever et al., 2014).


Adaptive networks offer a powerful and flexible framework for understanding the intricate dynamics of real-world systems that are inherently coupled. The bidirectional influence between the network structure and node states allows researchers to capture emergent behaviors that are otherwise difficult to model. This approach has been shown to enhance our understanding of many complex phenomena, such as the co-evolution of social ties and behaviors, the robustness of infrastructure networks, and the adaptive capabilities of biological systems.

**References**:
- Gross, T., & Sayama, H. (Eds.). (2009). *Adaptive Networks: Theory, Models and Applications*. Springer.
- Holme, P., & Newman, M. E. J. (2012). *Temporal Networks*. Springer.

- Gross, T., D'Lima, C. J. D., & Blasius, B. (2006). Epidemic dynamics on an adaptive network. *Physical Review Letters*, 96(20), 208701.</span>



## Adaptive Networks and Opinion Dynamics

Adaptive networks have proven to be a highly effective framework for studying opinion dynamics, particularly when the structure of social interactions evolves alongside individual opinions. This mutual adaptation is crucial for understanding how polarization, consensus, or radicalization may emerge in social systems. Models by Baumann and Starnini, as presented in *Physical Review Letters* (PRL) and *Physical Review X* (PRX), have significantly advanced our understanding of these phenomena by integrating homophily, social influence, and activity-driven interactions within adaptive networks.

In the model presented by Baumann et al. (2020), the dynamics of echo chambers and polarization are explored in a social network where individuals' opinions and their social connections evolve simultaneously. This model uses an activity-driven framework, where agents are characterized by heterogeneous activities that determine their propensity to interact. Connections are established not purely at random but are influenced by the similarity between agents' opinions, a concept known as homophily. The interactions are further modulated by the controversialness of the topic, which affects how strongly opinions influence each other. This model demonstrates how homophily and social influence can drive the system from a state of global consensus to radicalized or polarized states, depending on parameter values such as social influence strength and controversialness (Baumann et al., 2020, *PRL*).

The key equations governing the opinion dynamics and the adaptive nature of the network are:

**Equation 1**: The evolution of an agent's opinion $ x_i(t) $ is given by:
$$
\frac{d x_i}{d t} = -x_i + K \sum_{j=1}^{N} A_{ij}(t) \tanh(\alpha x_j),
$$
where $K$ represents the social interaction strength, $A_{ij}(t)$ is the temporal adjacency matrix defining the connections between agents at time $ t $, and $\alpha$ is a parameter controlling the influence of neighboring opinions.

**Equation 2**: The probability $ p_{ij} $ that an active agent $i$ will connect with a peer $j$ is modeled as:
$$
 p_{ij} = \frac{|x_i - x_j|^{-\beta}}{\sum_{k}|x_i - x_k|^{-\beta}},
$$
where the exponent $ \beta$ controls the strength of homophily, with higher values indicating a stronger preference for connecting with agents holding similar opinions.

The key insight from this work is that the combination of adaptive network structures and opinion dynamics leads to the formation of echo chambers—clusters of individuals with similar opinions that reinforce one another. These echo chambers are characterized by increased interaction among like-minded individuals, which leads to opinion polarization and makes it difficult for consensus to emerge. This emergent behavior reflects real-world social phenomena observed in online and offline communities, where individuals become increasingly isolated from differing viewpoints.

Starnini and colleagues (2021) expanded on this work in *Physical Review X* by focusing on the co-evolutionary dynamics between network topology and agent opinions in even greater detail. Their model incorporates both attractive and repulsive interactions, where agents not only strengthen connections with those who share similar views but may also sever ties or form negative relationships with those holding opposing views. This approach allows for the exploration of conditions under which polarization can lead to the coexistence of opposing factions or, alternatively, to the dominance of a single opinion. The results from these models provide insights into how social influence mechanisms, combined with adaptive connectivity, can lead to metastable polarized states or even rapid radicalization under certain conditions (Starnini et al., 2021, *PRX*).

These adaptive network models contribute significantly to our understanding of opinion dynamics by highlighting the importance of feedback loops between social interactions and individual beliefs. They illustrate how small changes in network connectivity or influence parameters can lead to vastly different outcomes, such as consensus, polarization, or fragmentation into isolated subgroups. The implications of these findings are broad, offering explanations for the persistence of political polarization, the spread of misinformation in online platforms, and the conditions that may lead to more cohesive or fragmented societies.

**References**:
- Baumann, F., Lorenz-Spreen, P., Sokolov, I. M., & Starnini, M. (2020). Modeling Echo Chambers and Polarization Dynamics in Social Networks. *Physical Review Letters*, 124(4), 048301.
- Starnini, M., Baumann, F., & Lorenz-Spreen, P. (2021). Co-evolutionary Dynamics of Opinions and Networks. *Physical Review X*, 11(3), 031044.</span>

### Homophily-based Activity driven model
```
def _homophily_based_AD(graph, gamma=2.0, m=10, activities=None, epsilon=1e-6):
    """
    Custom adaptation rule inspired by Starnini and Baumann's work on homophily and polarization.
    This rule aims to adapt the network structure by adding/removing edges based on homophily dynamics.

    Parameters:
        graph (dgl.DGLGraph): The DGL graph with node data attributes.
        gamma (float): Homophily exponent controlling the probability of connections based on opinion similarity.
        m (int): Number of edges to be created for each active node.
        activities (np.array): Array specifying the activity levels of each node.
        epsilon (float): Small value added to distances to prevent division by zero.
    """
    # print('Editing graph based on homophily')

    # Extract the beliefs (opinions) of each node from graph.ndata
    x = graph.ndata['beliefs'].cpu().numpy()

    # Removing all existing edges to reset the graph
    src, dst = graph.edges()
    graph.remove_edges(torch.arange(src.size(0)))

    # Number of nodes in the graph
    N = graph.num_nodes()

    # Function to determine which nodes are active
    def whos_active(activities):
        return np.where(np.random.uniform(0, 1, size=N) < activities)[0]

    # Function to sample targets based on homophily (1D opinions)
    def homophily_sampling_1d(i, x, gamma_hom, m, N, epsilon):
        distances_i = np.abs(x[i] - x) + epsilon
        distances_i[i] = 1  # Prevent division by zero for self

        p = (distances_i ** -gamma_hom)
        p[i] = 0  # No self-coupling

        p_sum = np.sum(p)
        if p_sum == 0:
            p[:] = 1 / (N - 1)  # If all probabilities are zero, use uniform distribution
        else:
            p /= p_sum

        return np.random.choice(range(N), m, replace=False, p=p)

    # Determine active nodes based on activities
    active_nodes = whos_active(activities)

    # Add edges based on homophily sampling
    for ai in active_nodes:
        targets_i = homophily_sampling_1d(ai, x, gamma, m, N, epsilon)
        for ti in targets_i:
            graph.add_edges(ai, ti)
            graph.add_edges(ti, ai)  # Add bidirectional edges

```

### Belief update rule
```
class BeliefUpdateRuleSigmaAD(torch.nn.Module):
    def __init__(self, alpha, K, dt):
        super(BeliefUpdateRuleSigmaAD, self).__init__()
        self.alpha = alpha
        self.K = K
        self.dt = dt

    def f_activity(self, sigma, x):
        return sigma - x

    def rkc_activity(self, f, dt, sigma, x):
        k1 = dt * f(sigma, x)
        k2 = dt * f(sigma, x + k1 / 2.)
        k3 = dt * f(sigma, x + k2 / 2.)
        k4 = dt * f(sigma, x + k3)
        return k1, k2, k3, k4

    def integration_step(self, f, dt, sigma, x):
        k1, k2, k3, k4 = self.rkc_activity(f, dt, sigma, x)
        return x + (1. / 6) * (k1 + 2 * k2 + 2 * k3 + k4)

    def forward(self, graph):
        """
        Custom belief update rule inspired by the sigma_inputs_AD function and equation 1 of the paper,
        adjusted for one-dimensional opinions.
        """
        adj_matrix = graph.adjacency_matrix().to_dense()
        beliefs = graph.ndata["beliefs"]

        Ax = torch.matmul(adj_matrix, beliefs)
        tanh_alpha_x = torch.tanh(self.alpha * Ax)

        new_beliefs = self.integration_step(self.f_activity, self.dt, self.K * tanh_alpha_x, beliefs)
        graph.ndata["beliefs"] = new_beliefs

```

    K_norm = one_minus_gamma / (high_a ** one_minus_gamma - low_a ** one_minus_gamma)
    a = (U * one_minus_gamma * K_norm ** -1 + low_a ** one_minus_gamma) ** (1. / one_minus_gamma)


    for t in range(timesteps):

        A = A_homophily(x, y, a, beta, N, m, delta)
        x_t[:, t] = x
        y_t[:, t] = y
        sigma_x = sigma_inputs_AD(A, x, y, K_x, alpha_x, delta)
        sigma_y = sigma_inputs_AD(A, y, x, K_y, alpha_y, delta)
        x = integration_step(f_activity, dt, sigma_x, x)
        y = integration_step(f_activity, dt, sigma_y, y)


    # return the results dict
    results = dict()
    results['t_array'] = t_array[:t]
    results['a'] = a
    results['time_to_switch'] = None
    results['x_final'] = x
    results['y_final'] = y
    results['sim_time'] = t
    results['x_t'] = x_t[:,:t]
    results['y_t'] = y_t[:,:t]

    return results

def scatter_hist(x_t, y_t, ax, ax_histx, ax_histy, params, p_label_top=False, plot_skewed_coords= False):

    def transform_to_polar_coordinates(x, y, N):

        r = np.zeros(N)
        phi = np.zeros(N)

        for i in range(N):

            r[i] = np.sqrt(x[i] ** 2 + y[i] ** 2)

            if y[i] < 0:
                phi[i] = -np.arccos(x[i] / r[i]) + 2 * np.pi
            else:
                phi[i] = np.arccos(x[i] / r[i])

        return r, phi

    x_t_cartesian, y_t_cartesian = x_t, y_t


    x_final = x_t_cartesian[:,-1]
    y_final = y_t_cartesian[:,-1]

    r, phi = transform_to_polar_coordinates(x_t[:, -1], y_t[:, -1], params['N'])


    for i in range(params['N']):
            ax.plot(x_t_cartesian[i, :], y_t_cartesian[i, :], lw=0.5, color='grey', alpha=0.5)

    ax.scatter(x_final, y_final, c=phi, s=10, cmap='hsv', alpha=0.75, zorder = 3, vmin = 0, vmax=2*np.pi)

    ax_histx.tick_params(axis="x", labelbottom=False)
    ax_histy.tick_params(axis="y", labelleft=False)
    ax_histx.hist(x_final, color = 'k', bins = 'auto', histtype='step', density = True)
    ax_histy.hist(y_final, color = 'k', bins = 'auto', histtype='step', orientation='horizontal', density =True)

    if p_label_top:
        ax_histx.set_ylabel(r'$P_1(x)$')
    ax_histy.set_xlabel(r'$P_2(x)$')
